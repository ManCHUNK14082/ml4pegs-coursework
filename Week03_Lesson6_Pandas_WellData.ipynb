## 1. Why Pandas? NumPy vs DataFrames

<div style="background: #f0f7ff; border-left: 5px solid #2d6a9f; padding: 16px 20px; border-radius: 0 8px 8px 0; margin: 12px 0;">

In Lesson 3 we used **NumPy arrays** to store a single log curve (e.g., one GR column).  
But real well data is **multi-column and labelled** — depths, GR, NPHI, RHOB, Formation names, Well IDs — all together.

**Pandas** gives us the **DataFrame**: a table with named columns, mixed data types, and powerful analysis tools. Think of it as Python's version of an Excel spreadsheet — but programmable.

| Feature | NumPy Array | Pandas DataFrame |
|---------|-------------|-----------------|
| Data shape | Single type, numerical | Mixed types (numbers + text) |
| Column names | None (index only) | Named columns |
| Row labels | Integer index | Custom index |
| Missing values | Awkward | Built-in (`NaN`) |
| GroupBy / aggregation | Manual loops | One line |
| Excel/CSV loading | External library | Built-in |

</div>

````python
# Import Pandas — standard alias is 'pd'
import pandas as pd
import numpy as np

print(f"Pandas version: {pd.__version__}")
print(f"NumPy  version: {np.__version__}")
````

<div style="background: #f0f7ff; border-left: 5px solid #2d6a9f; padding: 16px 20px; border-radius: 0 8px 8px 0; margin: 12px 0;">

## 5. Handling Missing Values

<div style="background:#f0f7ff;border-left:5px solid #2d6a9f;padding:14px 20px;border-radius:0 8px 8px 0;margin:10px 0;">

Real datasets almost always have gaps -- sensor failures, depth intervals where a tool was not run,
or values excluded during quality control. Pandas stores a missing value as **`NaN`** (Not a Number).

The standard workflow:

| Step | What you do | Key method |
|------|-------------|------------|
| **1. Detect** | Count NaNs per column | `df.isnull().sum()` |
| **2. Understand** | Decide what the gap means physically | domain knowledge |
| **3. Handle** | Drop rows/columns OR fill the gaps | `dropna()` / `fillna()` |

</div>

````python
# ── Step 1: Detect missing values ──────────────────────────────────────────
print("Missing value count per column:")
print(df.isnull().sum())

print("\nMissing as % of total rows:")
pct = (df.isnull().sum() / len(df) * 100).round(1)
print(pct)

print(f"\nTotal missing cells: {df.isnull().sum().sum()}")
print(f"Total cells:         {df.size}")
````

> **Why does RT_ohmm have the most gaps?**
> Resistivity tools are the most sensitive to borehole conditions (rugosity, fluid salinity, tool contact).
> It is common to see 10-15% missing resistivity readings in real wells. GR, being a passive nuclear
> measurement, is more robust and typically has fewer dropouts.

**Two main strategies:** drop the incomplete rows, or fill the gaps with an estimated value.
Which you choose depends on how many rows are affected and what the data will be used for.

````python
# ── Strategy A: dropna() -- remove every row that has ANY missing value ────
df_dropped = df.dropna()

print(f"Original rows:  {len(df)}")
print(f"After dropna(): {len(df_dropped)}")
print(f"Rows removed:   {len(df) - len(df_dropped)}")

# dropna(subset=[...]) limits the check to specific columns
df_dropped_gr = df.dropna(subset=["GR_API"])
print(f"\ndropna(subset=['GR_API']) rows: {len(df_dropped_gr)}")
print("Use subset= when only certain columns must be present.")
````

````python
# ── Strategy B: fillna() with column mean ───────────────────────────────────
# Suitable for log curves where the mean is a reasonable stand-in.
df_filled = df.copy()

log_cols = ["GR_API", "NPHI_frac", "RHOB_gcc", "RT_ohmm", "SW_frac"]
for col in log_cols:
    col_mean = df_filled[col].mean()
    df_filled[col] = df_filled[col].fillna(col_mean)
    print(f"{col:<12} -- filled {df[col].isnull().sum()} NaNs with mean = {col_mean:.3f}")

print(f"\nMissing values remaining: {df_filled.isnull().sum().sum()}")
````

````python
# ── Strategy C: forward fill -- best practice for well logs ─────────────────
# Well logs are continuous depth series. If a reading is missing at depth X,
# the reading just ABOVE (shallower) is the best geological estimate.
#
# ffill() propagates the last valid value downward.
# bfill() fills remaining gaps at the very top using the first valid value below.

df_clean = df.copy()
df_clean[log_cols] = df_clean.groupby("Well_ID")[log_cols].ffill()
df_clean[log_cols] = df_clean.groupby("Well_ID")[log_cols].bfill()

print("Forward fill applied per well (gaps filled within each well separately).")
print(f"Missing values remaining: {df_clean[log_cols].isnull().sum().sum()}")

# Use df_clean going forward
df = df_clean
print("\ndf is now the cleaned DataFrame -- ready for analysis.")
````

## 6. Selecting Columns and Rows

<div style="background: #f0f7ff; border-left: 5px solid #2d6a9f; padding: 16px 20px; border-radius: 0 8px 8px 0; margin: 12px 0;">

There are two main selection tools in Pandas:

| Method | Purpose | Example |
|--------|---------|---------|
| `df["col"]` | Select one column (returns a Series) | `df["GR_API"]` |
| `df[["col1","col2"]]` | Select multiple columns (returns a DataFrame) | `df[["GR_API","NPHI_frac"]]` |
| `df.iloc[row, col]` | Select by **integer position** | `df.iloc[0]` — first row |
| `df.loc[row, "col"]` | Select by **label** | `df.loc[0, "GR_API"]` |

</div>

## 7. Filtering Rows

<div style="background: #f0f7ff; border-left: 5px solid #2d6a9f; padding: 16px 20px; border-radius: 0 8px 8px 0; margin: 12px 0;">

Filtering in Pandas works exactly like NumPy boolean masks — you write a condition, get a True/False Series, then use it to select matching rows.

```python
df[df["GR_API"] < 75]      # all rows where GR < 75
```

You can combine conditions with `&` (AND) and `|` (OR). Always wrap each condition in parentheses.

</div>

````python
# Filter: only sand samples (GR < 75 API)
sand = df[df["GR_API"] < 75]
print(f"Sand samples: {len(sand)} rows")
print(sand[["Well_ID", "Depth_m", "Formation", "GR_API", "NPHI_frac"]].head(8))
````

````python
# Combined filter: good reservoir sand (GR < 75 AND porosity > 0.25)
good_reservoir = df[(df["GR_API"] < 75) & (df["NPHI_frac"] > 0.25)]
print(f"Good reservoir samples: {len(good_reservoir)} rows")
print(good_reservoir[["Well_ID","Formation","GR_API","NPHI_frac","SW_frac"]])
````

````python
# Filter by text column — select only Reservoir_1 rows
res1 = df[df["Formation"] == "Reservoir_1"]
print(f"Reservoir_1 samples: {len(res1)} rows")
print(res1[["Well_ID","Depth_m","GR_API","NPHI_frac"]].to_string(index=False))
````

````python
# Filter by Well ID — all data for Well_A only
well_a = df[df["Well_ID"] == "Well_A"]
print(f"Well_A has {len(well_a)} rows")
print(well_a)
````

> **Notice** — the same boolean mask pattern you learned with NumPy works directly in Pandas.  
> The key difference: Pandas preserves **column names** and **row labels**, making the output immediately interpretable.

<div style="background: #f0f7ff; border-left: 5px solid #2d6a9f; padding: 16px 20px; border-radius: 0 8px 8px 0; margin: 12px 0;">

You can add new columns to a DataFrame with a simple assignment.  
The right-hand side can be any vectorized expression — the same as NumPy.

````python
# Add Depth in feet
df["Depth_ft"] = df["Depth_m"] * 3.28084

# Add a lithology label based on GR cutoff
df["Lithology"] = df["GR_API"].apply(lambda gr: "Sand" if gr < 75 else "Shale")

# Add Hydrocarbon Pore Volume index: HCPV = NPHI × (1 - SW)
df["HCPV"] = df["NPHI_frac"] * (1 - df["SW_frac"])

print(df[["Well_ID","Depth_m","Depth_ft","GR_API","Lithology","NPHI_frac","SW_frac","HCPV"]].head(10))
````

## 9. GroupBy — Statistics by Zone or Formation

<div style="background: #f0f7ff; border-left: 5px solid #2d6a9f; padding: 16px 20px; border-radius: 0 8px 8px 0; margin: 12px 0;">

**GroupBy** splits the DataFrame into groups, applies a function to each group, and combines the results.  
It's the equivalent of a pivot table — one of the most powerful Pandas operations.

```python
df.groupby("Formation")["NPHI_frac"].mean()
```

Reads as: *"For each Formation, compute the mean NPHI."*

</div>

````python
# Mean porosity and GR by formation
formation_stats = df.groupby("Formation")[["GR_API","NPHI_frac","RHOB_gcc","SW_frac"]].mean().round(3)
print("=== Mean log values per Formation ===")
print(formation_stats)
````

````python
# Multiple aggregations at once: mean, min, max
gr_by_formation = df.groupby("Formation")["GR_API"].agg(["mean","min","max","std"]).round(1)
print("=== GR statistics per Formation ===")
print(gr_by_formation)
````

````python
# How many sand samples does each well have?
sand_count = df[df["GR_API"] < 75].groupby("Well_ID")["GR_API"].count()
print("=== Sand sample count per Well ===")
print(sand_count)

# Net-to-gross per well (sand rows / total rows, expressed as %)
total_by_well = df.groupby("Well_ID")["GR_API"].count()
sand_by_well  = df[df["GR_API"] < 75].groupby("Well_ID")["GR_API"].count()
ntg = (sand_by_well / total_by_well * 100).round(1)
print("\n=== Net-to-Gross per Well (%) ===")
print(ntg)
````

## 10. Plotting with Pandas

<div style="background: #f0f7ff; border-left: 5px solid #2d6a9f; padding: 16px 20px; border-radius: 0 8px 8px 0; margin: 12px 0;">

Pandas DataFrames have a built-in `.plot()` method that wraps Matplotlib.  
For quick exploratory plots it is faster than writing full Matplotlib code.

For well-log displays we still use Matplotlib directly (full control of axes).

</div>

````python
import matplotlib.pyplot as plt

# Cross-plot: GR vs NPHI coloured by formation (a classic petrophysical plot)
fig, ax = plt.subplots(figsize=(7, 5))

formations = df["Formation"].unique()
colors     = ["#2d6a9f", "#2ca87f", "#e07b39", "#9b59b6"]

for form, color in zip(formations, colors):
    subset = df[df["Formation"] == form]
    ax.scatter(subset["GR_API"], subset["NPHI_frac"],
               label=form, color=color, s=60, alpha=0.8, edgecolors="white", linewidth=0.5)

ax.set_xlabel("Gamma Ray (API)", fontsize=11)
ax.set_ylabel("Neutron Porosity (fraction)", fontsize=11)
ax.set_title("GR vs NPHI Cross-Plot — All Wells", fontsize=13, fontweight="bold")
ax.legend(fontsize=9)
ax.grid(linestyle="--", alpha=0.4)
plt.tight_layout()
plt.show()
````

````python
import matplotlib.pyplot as plt

# Multi-well GR log — each well as a separate colour
well_a = df[df["Well_ID"] == "Well_A"]
well_b = df[df["Well_ID"] == "Well_B"]
well_c = df[df["Well_ID"] == "Well_C"]

fig, ax = plt.subplots(figsize=(5, 9))

for well, color, label in [
    (well_a, "#2d6a9f", "Well A"),
    (well_b, "#2ca87f", "Well B"),
    (well_c, "#e07b39", "Well C"),
]:
    ax.plot(well["GR_API"], well["Depth_m"], color=color, linewidth=1.2,
            alpha=0.85, label=label)

ax.axvline(x=75, color="red", linestyle="--", linewidth=1.0, label="GR cutoff")
ax.invert_yaxis()
ax.set_xlabel("Gamma Ray (API)", fontsize=11)
ax.set_ylabel("Depth (m)", fontsize=11)
ax.set_title("GR Log — Three Wells", fontsize=12, fontweight="bold")
ax.legend(fontsize=9)
ax.grid(axis="x", linestyle="--", alpha=0.35)
plt.tight_layout()
plt.show()
````

````python
# Run the practice rebuild cell above first (loads df and adds Lithology/HCPV columns)

# ── Task 1: Net-to-Gross per well ─────────────────────────────────────────
# Net-to-Gross = fraction of depth that is sand (GR < 75 API)

# Your code here:
# print(df[["GR_API","Lithology","NPHI_frac"]].tail())
df = pd.read_csv("well_Log_data.csv")

df["Lithology"] = df["GR_API"].apply(lambda gr: "Sand" if gr < 75 else "Shale")

net_to_gross = df[df["GR_API"] < 75].groupby("Well_ID")["Depth_m"].count()
print(max(net_to_gross))

Well_C = df[df["Well_ID"] == 'Well_C']
mean_RHOB = df.groupby("Formation")["RHOB_gcc"].mean().sort_values(ascending=False).round(4)
print(Well_C[["GR_API","NPHI_frac","Lithology"]].to_string(index=False))
print(f"Lowest density:",min(mean_RHOB))
````

Using the multi-well dataset:

### 📝 Task 3 — Student Grade Tracker

Create a DataFrame representing **20 students** with columns:
- `Name` (any names — you can use placeholder names like "Student_01")
- `Course` (pick from: "Physics", "Chemistry", "Mathematics", "Biology")
- `Midterm` (score out of 100)
- `Final` (score out of 100)

Then:
1. Add a `Total` column = average of Midterm and Final.
2. Add a `Grade` column using `.apply()`: A (≥ 80), B (65–79), C (< 65).
3. Use GroupBy to find the **mean Total score per Course**.
4. Print the 5 highest-scoring students.

````python
import pandas as pd
import numpy as np

np.random.seed(7)
n = 20
courses = ["Physics", "Chemistry", "Mathematics", "Biology"]

students = pd.DataFrame({
    "Name":     [f"Student_{i+1:02d}" for i in range(n)],
    "Course":   np.random.choice(courses, n),
    "Midterm":  np.random.randint(45, 98, n),
    "Final":    np.random.randint(48, 99, n),
})

# ── Your code here ────────────────────────────────────────────────────────────
students["Total"] = (students["Midterm"] + students["Final"]) / 2

def assign_grade(score):
    if score >= 80: return "A"
    elif score >= 65: return "B"
    else: return "C"

students["Grade"] = students["Total"].apply(assign_grade)

print("=== Mean score per Course ===")
print(students.groupby("Course")["Total"].mean().round(1))

print("\n=== Top 5 Students ===")
print(students.nlargest(5, "Total")[["Name","Course","Midterm","Final","Total","Grade"]])
````

### 📝 Task 4 — Movie Ratings Analysis

You have a small movie ratings dataset. Build it as a DataFrame and perform a GroupBy analysis.

**Data:**

| Movie | Genre | Rating | Year |
|-------|-------|--------|------|
| Inception | Sci-Fi | 8.8 | 2010 |
| The Dark Knight | Action | 9.0 | 2008 |
| Interstellar | Sci-Fi | 8.6 | 2014 |
| Avengers: Endgame | Action | 8.4 | 2019 |
| Parasite | Drama | 8.6 | 2019 |
| The Shawshank Redemption | Drama | 9.3 | 1994 |
| Arrival | Sci-Fi | 7.9 | 2016 |
| Mad Max: Fury Road | Action | 8.1 | 2015 |
| Moonlight | Drama | 7.4 | 2016 |
| Dune | Sci-Fi | 8.0 | 2021 |

1. Create the DataFrame.
2. Find the **mean rating per Genre**.
3. Filter to movies rated **≥ 8.5**.
4. Add a `Decade` column: `(df["Year"] // 10) * 10`.
5. Print the mean rating per Decade.

````python
import pandas as pd

movies = pd.DataFrame({
    "Movie":  ["Inception","The Dark Knight","Interstellar","Avengers: Endgame",
               "Parasite","The Shawshank Redemption","Arrival","Mad Max: Fury Road",
               "Moonlight","Dune"],
    "Genre":  ["Sci-Fi","Action","Sci-Fi","Action","Drama","Drama","Sci-Fi","Action","Drama","Sci-Fi"],
    "Rating": [8.8, 9.0, 8.6, 8.4, 8.6, 9.3, 7.9, 8.1, 7.4, 8.0],
    "Year":   [2010, 2008, 2014, 2019, 2019, 1994, 2016, 2015, 2016, 2021],
})

# ── Your code here ────────────────────────────────────────────────────────────
print("=== Mean Rating per Genre ===")
print(movies.groupby("Genre")["Rating"].mean().round(2))

print("\n=== Movies rated ≥ 8.5 ===")
print(movies[movies["Rating"] >= 8.5][["Movie","Genre","Rating"]])

movies["Decade"] = (movies["Year"] // 10) * 10
print("\n=== Mean Rating per Decade ===")
print(movies.groupby("Decade")["Rating"].mean().round(2))
````
