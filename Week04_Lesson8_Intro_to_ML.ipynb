## 2. What is Machine Learning?

<div style="background: #f0f7ff; border-left: 5px solid #2d6a9f; padding: 16px 20px; border-radius: 0 8px 8px 0; margin: 12px 0;">

**Traditional programming:**  
You write rules → the computer follows them → you get output.

**Machine Learning:**  
You give the computer data + answers → the computer **learns the rules itself** → it can then predict answers for new data.

Machine learning is a branch of Artificial Intelligence (AI) where algorithms **learn patterns from data** without being explicitly programmed for each rule.

</div>

### The Three Learning Paradigms

| Paradigm | How It Learns | Petroleum Example |
|----------|--------------|-------------------|
| **Supervised Learning** | Learns from labelled examples (input + correct answer) | Predict porosity from GR and RHOB |
| **Unsupervised Learning** | Finds hidden patterns in unlabelled data | Group wells into clusters by log character |
| **Reinforcement Learning** | Learns by trial and error, rewarded for good actions | Optimise drilling parameters in real time |

> **This lesson focuses on Supervised Learning** -- the most common type in the energy industry. You provide the model with pairs of (input, target) and it learns the mapping between them.

### Supervised Learning -- Two Types

<div style="background: #f0f7ff; border-left: 5px solid #2d6a9f; padding: 16px 20px; border-radius: 0 8px 8px 0; margin: 12px 0;">

| Type | Output | Example |
|------|--------|---------|
| **Regression** | A continuous number | Predict RHOB from GR (any decimal value) |
| **Classification** | A category/label | Classify a formation as Sand or Shale |

In this lesson you will build a **regression** model.

## 3. The 5-Step ML Workflow

Every ML project -- whether in academia or industry -- follows this same sequence:

<div style="background: #f0f7ff; border-left: 5px solid #2d6a9f; padding: 16px 20px; border-radius: 0 8px 8px 0; margin: 12px 0;">

| Step | Action | In This Lesson |
|------|--------|----------------|
| **1. Collect & load data** | Get your dataset into Python | Load `well_log_data.csv` |
| **2. Prepare features** | Select inputs and target; split into train/test sets | Choose RHOB as input, NPHI as target |
| **3. Choose a model** | Pick an algorithm suited to the problem | `LinearRegression` from scikit-learn |
| **4. Train the model** | Fit the model to the training data | `model.fit(X_train, y_train)` |
| **5. Evaluate** | Measure how well it predicts on unseen data | R² and RMSE on the test set |

</div>

> **Key idea -- train/test split:**  
> We hold out a portion of the data (typically 20%) as a **test set** that the model never sees during training. This simulates predicting on genuinely new data and gives an honest measure of performance.

## 4. Step 1 & 2 -- Prepare Your Data

We use our familiar `well_log_data.csv`. The model will learn to **predict NPHI (neutron porosity) from RHOB (bulk density)** -- a classic petrophysical relationship: denser rock = less pore space = lower neutron porosity.

```python
# Import all libraries we need
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error, r2_score

print("All libraries loaded successfully!")
```

```python
# Step 1: Load and clean data (same as previous lessons)
df = pd.read_csv("well_log_data.csv")

log_cols = ["GR_API", "NPHI_frac", "RHOB_gcc", "RT_ohmm", "SW_frac"]
df[log_cols] = df.groupby("Well_ID")[log_cols].ffill()
df[log_cols] = df.groupby("Well_ID")[log_cols].bfill()

print("Dataset loaded. Shape:", df.shape)
print("Missing values remaining:", df[log_cols].isnull().sum().sum())
df[["Well_ID", "Depth_m", "RHOB_gcc", "NPHI_frac"]].head(6)
```

### Features and Target

<div style="background: #f0f7ff; border-left: 5px solid #2d6a9f; padding: 16px 20px; border-radius: 0 8px 8px 0; margin: 12px 0;">

In ML, we use specific names:

| Term | Meaning | In This Model |
|------|---------|---------------|
| **Feature** (X) | The input column(s) the model uses to predict | `RHOB_gcc` -- bulk density |
| **Target** (y) | The column we want to predict | `NPHI_frac` -- neutron porosity |
| **Training set** | Data the model learns from (80%) | 576 rows |
| **Test set** | Data we evaluate on -- model never sees this during training (20%) | 144 rows |

</div>

```python
# Step 2a: Explore the relationship visually before modelling
fig, ax = plt.subplots(figsize=(7, 5))
ax.scatter(df["RHOB_gcc"], df["NPHI_frac"], alpha=0.25, s=10, color="#2d6a9f")
ax.set_xlabel("RHOB -- Bulk Density (g/cc)")
ax.set_ylabel("NPHI -- Neutron Porosity (frac)")
ax.set_title("Density vs Porosity Crossplot -- All Wells")
ax.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
print("Correlation coefficient:", round(df["RHOB_gcc"].corr(df["NPHI_frac"]), 3))
```

> **What the crossplot shows:** Higher density (RHOB) corresponds to lower porosity (NPHI) in most rows -- a negative correlation. This is the physical basis of the density-porosity transform used in petrophysics every day. ML will learn this relationship from the data without us writing the equation.

```python
# Step 2b: Select feature and target, then split into train and test sets

# X must be a 2D array (n_samples, n_features); reshape single column with [[ ]]
X = df[["RHOB_gcc"]]      # feature -- 2D DataFrame
y = df["NPHI_frac"]       # target  -- 1D Series

# Split: 80% training, 20% testing
# random_state=42 ensures the same split every time you run the cell
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, random_state=42
)

print(f"Total rows     : {len(df)}")
print(f"Training rows  : {len(X_train)}  ({len(X_train)/len(df)*100:.0f}%)")
print(f"Test rows      : {len(X_test)}  ({len(X_test)/len(df)*100:.0f}%)")
```

---
## 5. Step 3 & 4 -- Build and Train a Model

### What is Linear Regression?

<div style="background: #f0f7ff; border-left: 5px solid #2d6a9f; padding: 16px 20px; border-radius: 0 8px 8px 0; margin: 12px 0;">

Linear Regression fits the best straight line through your data:

$$\hat{y} = m \cdot x + b$$

- $\hat{y}$ -- predicted value (NPHI)
- $x$ -- input value (RHOB)
- $m$ -- slope (how much NPHI changes per unit RHOB)
- $b$ -- intercept (NPHI when RHOB = 0)

**Training** = finding the values of $m$ and $b$ that minimise the error between predictions and true values. scikit-learn does all this in one line: `model.fit(X_train, y_train)`.

</div>

```python
# Step 3: Create the model object
model = LinearRegression()

# Step 4: Train -- the model finds the best slope (m) and intercept (b)
model.fit(X_train, y_train)

print("Model trained!")
print(f"  Slope (m)     : {model.coef_[0]:.4f}   -- NPHI changes by this much per 1 g/cc increase in RHOB")
print(f"  Intercept (b) : {model.intercept_:.4f}  -- NPHI predicted when RHOB = 0")
print(f"\n  Equation: NPHI = {model.coef_[0]:.4f} * RHOB + {model.intercept_:.4f}")
```

```python
# Make predictions on the TEST set (data the model has never seen)
y_pred = model.predict(X_test)

# Show a few predictions vs actual values
comparison = pd.DataFrame({
    "RHOB_gcc": X_test["RHOB_gcc"].values,
    "Actual_NPHI": y_test.values,
    "Predicted_NPHI": y_pred.round(4),
    "Error": (y_pred - y_test.values).round(4)
})
print("Sample predictions vs actual values:")
print(comparison.head(10).to_string(index=False))
```

```python
# Visualise the fitted line on the training data
rhob_range = np.linspace(df["RHOB_gcc"].min(), df["RHOB_gcc"].max(), 100).reshape(-1, 1)
nphi_line = model.predict(rhob_range)

fig, ax = plt.subplots(figsize=(7, 5))
ax.scatter(X_train["RHOB_gcc"], y_train, alpha=0.2, s=10, color="#2d6a9f", label="Training data")
ax.scatter(X_test["RHOB_gcc"], y_test, alpha=0.4, s=14, color="#2ca87f", label="Test data")
ax.plot(rhob_range, nphi_line, color="#c0392b", linewidth=2, label="Model prediction")
ax.set_xlabel("RHOB (g/cc)")
ax.set_ylabel("NPHI (frac)")
ax.set_title("Linear Regression -- NPHI Predicted from RHOB")
ax.legend(fontsize=9)
ax.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

---
## 6. Step 5 -- Evaluate the Model

Training a model is easy. Knowing **how good it is** is the important part.

<div style="background: #f0f7ff; border-left: 5px solid #2d6a9f; padding: 16px 20px; border-radius: 0 8px 8px 0; margin: 12px 0;">

| Metric | Formula | Interpretation |
|--------|---------|----------------|
| **R² (R-squared)** | 1 - SS_res / SS_tot | How much of the variation in y the model explains. 1.0 = perfect; 0 = no better than the mean |
| **MSE** | mean of (y_pred - y_actual)² | Average squared error -- penalises large mistakes heavily |
| **RMSE** | sqrt(MSE) | Same units as the target variable -- easier to interpret |

> **Rule of thumb:** R² above 0.7 is generally considered a useful model. RMSE should be small relative to the range of the target variable.

</div>

```python
# Calculate evaluation metrics on the test set
r2   = r2_score(y_test, y_pred)
mse  = mean_squared_error(y_test, y_pred)
rmse = np.sqrt(mse)
nphi_range = y_test.max() - y_test.min()

print("=== Model Performance on Test Set ===")
print(f"  R-squared (R2) : {r2:.4f}")
print(f"  MSE            : {mse:.6f}")
print(f"  RMSE           : {rmse:.4f}  (NPHI units: frac)")
print(f"  NPHI range     : {nphi_range:.4f}  (frac)")
print(f"  RMSE / range   : {rmse/nphi_range*100:.1f}% of total range")

if r2 > 0.7:
    print("\n  Assessment: Good fit -- the model explains most of the variation.")
elif r2 > 0.4:
    print("\n  Assessment: Moderate fit -- there is real signal but also unexplained scatter.")
else:
    print("\n  Assessment: Weak fit -- a single feature is not enough to explain NPHI variation.")
```

```python
# Residual plot -- check if errors are random (they should be for a good model)
residuals = y_test.values - y_pred

fig, axes = plt.subplots(1, 2, figsize=(11, 4))

# Left: Actual vs Predicted
axes[0].scatter(y_test, y_pred, alpha=0.3, s=12, color="#2d6a9f")
min_val = min(y_test.min(), y_pred.min())
max_val = max(y_test.max(), y_pred.max())
axes[0].plot([min_val, max_val], [min_val, max_val], color="#c0392b", linewidth=1.5, linestyle="--", label="Perfect prediction")
axes[0].set_xlabel("Actual NPHI (frac)")
axes[0].set_ylabel("Predicted NPHI (frac)")
axes[0].set_title("Actual vs Predicted")
axes[0].legend(fontsize=8)
axes[0].grid(True, alpha=0.3)

# Right: Residuals
axes[1].scatter(y_pred, residuals, alpha=0.3, s=12, color="#2ca87f")
axes[1].axhline(0, color="#c0392b", linewidth=1.5, linestyle="--")
axes[1].set_xlabel("Predicted NPHI (frac)")
axes[1].set_ylabel("Residual (Actual -- Predicted)")
axes[1].set_title("Residual Plot")
axes[1].grid(True, alpha=0.3)

plt.suptitle(f"Model Evaluation  |  R² = {r2:.3f}  |  RMSE = {rmse:.4f}", fontsize=10)
plt.tight_layout()
plt.show()
```

### Interpreting the Results

<div style="background: #f0f7ff; border-left: 5px solid #2d6a9f; padding: 16px 20px; border-radius: 0 8px 8px 0; margin: 12px 0;">

**Actual vs Predicted plot:** If all points fell exactly on the dashed diagonal, the model would be perfect. Points clustered around the line mean the model is useful. Wide scatter means the model is missing information.

**Residual plot:** Residuals (errors) should be randomly scattered around zero with no pattern. A curved or funnel-shaped pattern means the linear model does not fully capture the relationship -- a more complex model may be needed.

**Why is R² not 1.0?** NPHI depends on more than just RHOB -- it also responds to fluid type, clay content, and gas. A single feature can never fully explain it. In later lessons you will use **multiple features** to improve predictions.

</div>
