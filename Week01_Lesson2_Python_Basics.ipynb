## 2. Variables

<div style="background: #f0f7ff; border-left: 5px solid #2d6a9f; padding: 16px 20px; border-radius: 0 8px 8px 0; margin: 12px 0;">

### What is a Variable?

> **A variable is a container used to store information.**

Think of it like a labelled box: you put a value inside and give the box a name so you can find it later.

In petroleum engineering, we work with quantities like:
- Pressure (psi)
- Temperature (°C or °F)
- Depth (m or ft)
- Porosity (fraction)

Variables let us store these values and use them in calculations.

</div>

````python
pressure = 3500
temperature = 120
depth = 2500

print(pressure)
print(temperature)
print(depth)
````

### Data Types — Three You Need to Know

| Type | Description | Example |
|------|-------------|--------|
| **Integer** (`int`) | Whole numbers, no decimal point | `depth = 2500` |
| **Float** (`float`) | Numbers with a decimal point | `porosity = 0.18` |
| **String** (`str`) | Text, wrapped in quotes | `well_name = "Well_A"` |

````python
# Demonstrating data types
well_name = "Well_A"   # string — text
porosity  = 0.18       # float  — decimal number
zone      = 3          # int    — whole number

print(well_name)
print(porosity)
print(zone)

# You can check the type of any variable with type()
print(type(well_name))
print(type(porosity))
print(type(zone))
````

````python
well_name  ="Well_B"
porosity   = 0.18
zone       = 3

print(well_name)
print(porosity)
print(zone)

print(type(well_name))
print(type(porosity))
print(type(zone))
````

### Printing Variables with Labels (f-strings)

It is better practice to label your output so it is clear what each number means.  
Python provides **f-strings** for this — put an `f` before the quote, then use `{}` to insert a variable:

````python
# f-strings make output more readable
pressure    = 3500
temperature = 120
depth       = 2500

print(f"Reservoir pressure:    {pressure} psi")
print(f"Reservoir temperature: {temperature} °C")
print(f"Well depth:            {depth} m")
````

````python
pressure     = 3500
temperature  = 120
depth         = 2500

print(f"Reservoir pressure:{pressure}psi")
print(f"Reservoir temperature:{temperature}°C")
print(f"Reservoir depth:{depth}m")
````

## 3. Basic Mathematical Operations

<div style="background: #f0f7ff; border-left: 5px solid #2d6a9f; padding: 16px 20px; border-radius: 0 8px 8px 0; margin: 12px 0;">

Python works as a powerful calculator. You can use standard arithmetic operators:

| Operation | Symbol | Example | Result |
|-----------|--------|---------|--------|
| Addition | `+` | `10 + 5` | `15` |
| Subtraction | `-` | `10 - 5` | `5` |
| Multiplication | `*` | `10 * 5` | `50` |
| Division | `/` | `10 / 5` | `2.0` |
| Power / Exponent | `**` | `2 ** 3` | `8` |
| Modulo (remainder) | `%` | `10 % 3` | `1` |

</div>

````python
# Basic arithmetic operators
a = 10
b = 5

print(f"Addition:       {a} + {b} = {a + b}")
print(f"Subtraction:    {a} - {b} = {a - b}")
print(f"Multiplication: {a} * {b} = {a * b}")
print(f"Division:       {a} / {b} = {a / b}")
````

````python
# Additional operators
print(f"Power:   2 ** 3 = {2 ** 3}")   # 2 to the power of 3
print(f"Modulo: 10 % 3  = {10 % 3}")   # remainder when 10 is divided by 3
````

### 🛢️ Engineering Example — Velocity Calculation

In seismic surveys, wave **velocity** is fundamental to locating subsurface reflectors.

$$\text{Velocity} = \frac{\text{Distance}}{\text{Time}}$$

If a seismic wave travels **1500 m** in **3 seconds**, what is its velocity?

````python
# Seismic wave velocity calculation
distance = 1500    # metres
time     = 3       # seconds

velocity = distance / time

print(f"Distance: {distance} m")
print(f"Time:     {time} s")
print(f"Velocity: {velocity} m/s")
````

> **Notice:** Python did exactly what a formula does — once you store values in variables, you can write equations using those variable names. This is the foundation of every engineering calculation you will write in this course.

### 🛢️ Engineering Example 2 — Pressure Gradient

The **hydrostatic pressure** at a given depth is:

$$P = \rho \cdot g \cdot h$$

Where:
- $\rho$ = fluid density (kg/m³)
- $g$ = gravitational acceleration (9.81 m/s²)
- $h$ = depth (m)

````python
# Hydrostatic pressure calculation
rho   = 1025     # seawater density in kg/m³
g     = 9.81     # gravitational acceleration in m/s²
h     = 500      # depth in metres

pressure_pa  = rho * g * h          # pressure in Pascals
pressure_bar = pressure_pa / 1e5    # convert to bar

print(f"Depth:            {h} m")
print(f"Fluid density:    {rho} kg/m³")
print(f"Pressure:         {pressure_pa:,.0f} Pa")
print(f"Pressure:         {pressure_bar:.2f} bar")
````

## 4. Lists in Python

<div style="background: #f0f7ff; border-left: 5px solid #2d6a9f; padding: 16px 20px; border-radius: 0 8px 8px 0; margin: 12px 0;">

### What is a List?

> **A list stores multiple values in a single variable.**

In geoscience, we rarely work with a single measurement. We work with:
- A series of seismic velocities at different depths
- Temperature readings over time
- Pressure values across multiple wells

A **list** is Python's way of holding a sequence of values together.

</div>

````python
# Creating a list — square brackets [], values separated by commas
velocities = [1500, 1800, 2100, 2500]

print(velocities)
print(f"Type: {type(velocities)}")
````

### Accessing Items — Index (Position)

Each item in a list has an **index** — its position number.  

⚠️ **Python counting starts from 0**, not 1!

```
velocities = [1500, 1800, 2100, 2500]
Index:           0     1     2     3
```

````python
velocities = [1500, 1800, 2100, 2500]

print(f"First value  (index 0): {velocities[0]} m/s")
print(f"Second value (index 1): {velocities[1]} m/s")
print(f"Third value  (index 2): {velocities[2]} m/s")
print(f"Last value   (index -1): {velocities[-1]} m/s")  # -1 always gives the last item
````

### Modifying a List

You can change any item in a list by assigning a new value to its index:

````python
velocities = [1500, 1800, 2100, 2500]
print(f"Before: {velocities}")

velocities[1] = 1900    # update the second value
print(f"After:  {velocities}")
````

### List Length and Useful Operations

````python
velocities = [1500, 1800, 2100, 2500]

print(f"Number of items: {len(velocities)}")
print(f"Minimum:         {min(velocities)} m/s")
print(f"Maximum:         {max(velocities)} m/s")
print(f"Sum:             {sum(velocities)} m/s")
print(f"Average:         {sum(velocities)/len(velocities):.1f} m/s")
````

### Adding Items to a List

````python
velocities = [1500, 1800, 2100, 2500]
print(f"Before: {velocities}")

velocities.append(2800)    # .append() adds one item to the end
print(f"After:  {velocities}")
print(f"New length: {len(velocities)}")
````

### Lists Can Hold Mixed Types

A list can contain strings, numbers, and even other lists:

````python
# A well record: name, depth (m), porosity, is_producer
well_info = ["Well_A", 2500, 0.18, True]

print(f"Well name:   {well_info[0]}")
print(f"Depth:       {well_info[1]} m")
print(f"Porosity:    {well_info[2]*100:.0f}%")
print(f"Is producer: {well_info[3]}")
````
